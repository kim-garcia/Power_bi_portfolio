```python
import pandas as pd
import numpy as np
from typing import Dict, Tuple

print("Calculadora de Retenciones")
print("=" * 50)
```

```python
# AFP
afp_empleado = 0.0725
afp_empleador = 0.0775

# tasa ISSS
isss_empleado = 0.03
isss_empleador = 0.075
techo_isss = 1000.00
ISSS_MAXIMO_EMPLEADO = 30.00

# tabla isr
tabla_isr =  [
    {"desde": 0.01, "hasta": 472.00, "porcentaje": 0.00, "cuota_fija": 0.00, "sobre_exceso": 0.01},
    {"desde": 472.01, "hasta": 895.24, "porcentaje": 0.10, "cuota_fija": 17.67, "sobre_exceso": 472.01},
    {"desde": 895.25, "hasta": 2038.10, "porcentaje": 0.20, "cuota_fija": 60.00, "sobre_exceso": 895.25},
    {"desde": 2038.11, "hasta": float('inf'), "porcentaje": 0.30, "cuota_fija": 288.57, "sobre_exceso": 2038.11}
]

print("Constantes definidas:")
print(f"AFP Empleado: {afp_empleado*100:.2f}%")
print(f"AFP Empleador: {afp_empleador*100:.2f}%")
print(f"ISSS Empleado: {isss_empleado*100:.1f}%")
print(f"ISSS Empleador: {isss_empleador*100:.2f}%")
print(f"Techo ISSS: ${techo_isss:.2f}")
print(f"ISSS máximo empleado: ${ISSS_MAXIMO_EMPLEADO:.2f}")
```

```python
def calcular_afp_empleado(salario: float) -> float:
    """
    Calcula la contribución AFP del empleado (7.25%)

    Args:
        salario (float): Salario mensual bruto

    Returns:
        float: Monto de AFP empleado
    """
    return salario * afp_empleado

def calcular_afp_empleador(salario: float) -> float:
    """
    Calcula la contribución AFP del empleador (7.75%)

    Args:
        salario (float): Salario mensual bruto

    Returns:
        float: Monto de AFP empleador
    """
    return salario * afp_empleador

# Prueba de las funciones AFP
salario_prueba = 1000.00
print(f"Prueba con salario de ${salario_prueba:.2f}:")
print(f"AFP Empleado: ${calcular_afp_empleado(salario_prueba):.2f}")
print(f"AFP Empleador: ${calcular_afp_empleador(salario_prueba):.2f}")
```

```python
def calcular_isss_empleado(salario: float) -> float:
    """
    Calcula la contribución ISSS del empleado (3% con techo de $1000)
    Si el salario > $1000, el ISSS empleado es $30.00 máximo

    Args:
        salario (float): Salario mensual bruto

    Returns:
        float: Monto de ISSS empleado
    """
    if salario > techo_isss:
        return ISSS_MAXIMO_EMPLEADO
    else:
        return salario * isss_empleado
def calcular_isss_empleador(salario: float) -> float:
    """
    Calcula la contribución ISSS del empleador (7.50%)

    Args:
        salario (float): Salario mensual bruto
    Returns:
        float: Monto de ISSS empleador
    """
    return salario * isss_empleador

# Prueba de las funciones ISSS con diferentes salarios
salarios_prueba = [800.00, 1000.00, 1500.00, 2000.00]

print("Pruebas de ISSS con diferentes salarios:")
print("-" * 40)
for salario in salarios_prueba:
    isss_emp = calcular_isss_empleado(salario)
    isss_empleador = calcular_isss_empleador(salario)
    print(f"Salario: ${salario:.2f}")
    print(f"  ISSS Empleado: ${isss_emp:.2f}")
    print(f"  ISSS Empleador: ${isss_empleador:.2f}")
    print()
```
